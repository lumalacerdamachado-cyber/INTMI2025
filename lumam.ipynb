##Operadores

adição +

subtração -

divisão /

multiplicação *

resto da divisão %

divisão inteira //

exponenciação **

```python
numero1 = 10
numero2 = 5

print("A soma é igual a:", numero1 + numero2)
print("A subtração soma é igual a:", numero1 - numero2)
print("A divisão é igual a:", numero1 / numero2)
print("A multiplicação  é igual a:", numero1 * numero2)
print("o resto da divisão é igual a:", numero1 % numero2)
print("A divisão inteira é igual a:", numero1 // numero2)
print("A  exponenciação é igual a:", numero1 ** numero2)
```

##Estrutura Condicional

if - se

elif - senão se

else - senão

```python
nota = float(input("Digite a sua nota: "))
if (nota >= 7):
  print("Aprovado")
elif (nota >= 5):
  print("Recuperação")
else:
  print("Reprovado")
```

4 - Crie um algoritmo que leia a sua idade e dependendo mostre as frases na tela:

até 12 anos - Você é uma Criança!

13 - 18   - Você é Adolescente!

19 - 60   - Você é Adulto!

acima de 60 - Você é Idoso!

```python
idade = int(input("Digite a sua idade"))
if (idade <=12):
  print("Você é uma criança")
elif(idade >=13 and idade <=18):
  print("Você é Adolescente")
elif(idade >=19 and idade <=60):
  print("Você é Adulto")
else:
  print("Você é Idoso")
```
